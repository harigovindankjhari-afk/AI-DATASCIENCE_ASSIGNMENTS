```python
import numpy as np
import pandas as pd
from sklearn.datasets import load_iris
from sklearn.cluster import AgglomerativeClustering
from scipy.cluster.hierarchy import dendrogram,linkage
import matplotlib.pyplot as plt

```

```python
iris=load_iris()
```

```python
x=iris.data
y=iris.target
```

```python
x
```

```python
y
```

```python
print(x.shape)
```

```python
linked = linkage(x,method='ward')
plt.figure(figsize=(10,5))
dendrogram(linked)
plt.title('Dendrogram - Iris Data Set')
plt.xlabel("Sample")
plt.ylabel("Distance")

```

```python
model=AgglomerativeClustering(n_clusters=3,linkage='ward')
labels=model.fit_predict(x)
print(labels[:10])
```

```python
model
```

```python
#Visualizing the clustering
```

```python
plt.scatter(x[:,2],x[:,3],c=labels,cmap='rainbow')
plt.xlabel('Petal length')
plt.ylabel('Petal width')
plt.title("Agglomerative Clustering(iris)")
plt.show()
```

```python
from sklearn.metrics import accuracy_score
print('ActualLabels',y[:10])
print('PredictedLabel',labels[:10])
```

```python
import pandas as pd
import numpy as np
from sklearn.datasets import load_iris
from sklearn.cluster import DBSCAN
from sklearn.preprocessing import StandardScaler
```

```python
iris=load_iris()
x=iris.data
y=iris.target
```

```python
x
```

```python
y
```

```python
scaler=StandardScaler()
x_scaled= scaler.fit_transform(x)
x_scaled
```

```python
dbscan = DBSCAN(eps=0.8,min_samples=5)
labels=dbscan.fit_predict(x_scaled)
print("ClusteredLabels:",labels)
```

```python
y
```

```python
plt.scatter(x_scaled[:,2],x_scaled[:,3],c=labels,cmap='rainbow')
plt.xlabel('Petal Length')
plt.ylabel('Petal Width')
plt.title('DBSCAN Clustering(Iris)')
plt.show()
```

```python
n_clusters=len(set(labels))-(1 if -1 in labels else 0)
n_noise=list(labels).count(-1)
print('Estimated number of clusters:',n_clusters)
print('Estimated number of noise points:',n_noise)
```
