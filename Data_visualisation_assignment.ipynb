```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
df=pd.read_csv("tested.csv")
df.head()
```

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
sns.countplot(x="Survived",data=df,palette="colorblind")
plt.title("count plot of passenger details")
plt.show()
```

```python
sns.swarmplot(x="Survived",y="Age",data=df,palette="colorblind")
plt.title("Swarm plot")
plt.show()
```

```python
import matplotlib.pyplot as plt
passengers=[100,200,300,400,500,600,700]
Age=[10,20,30,40,50,60,70]
colors=['gold','lightblue','lightgreen','pink','red','orange','yellow']
explode=(0,0,0,0,0,0,0)
plt.pie(
    passengers,
    labels=Age,
    explode=explode,
    autopct='%1.1f%%',
    shadow=True,
    startangle=140,

)

```

```python
num_df=df.select_dtypes(include=["int64","float64"])
corr=num_df.corr()
plt.figure(figsize=(10,6))
sns.heatmap(corr,annot=True,cmap="coolwarm")
plt.title("correlation Heatmap of Titanic Dataset")
plt.show()
```

```python
sns.violinplot(x="Survived",y="Fare",data=df,palette='colorblind')
plt.title("Violinplot")
plt.show()
```

```python
sns.scatterplot(x="Age",y="Fare",data=df,palette="colorblind")
plt.title("Scatterplot")
plt.show()
```
