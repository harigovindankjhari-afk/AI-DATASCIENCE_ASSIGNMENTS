```python
import requests
from bs4 import BeautifulSoup
url="https://toscrape.com"
response=requests.get(url)
html_content=response.text
soup=BeautifulSoup(html_content,"html.parser")
print(soup.prettify())
```

```python
h=soup.find("h1")
print(h.text)
```
