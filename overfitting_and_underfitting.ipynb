```python
import pandas as pd
from sklearn.linear_model import LinearRegression

# 1. Create the dataset with area, bedroom, age, and price
data_house = pd.DataFrame({
    'area': [1500, 1800, 2400, 3000, 1200, 3500],
    'bedroom': [3, 3, 4, 4, 2, 5],
    'age': [10, 15, 5, 8, 20, 2],
    'price': [400000, 450000, 580000, 650000, 320000, 780000]
})
data_house
```

```python
x=data_house.drop('price',axis=1)
y=data_house['price']
```

```python
print(x)
print(y)
```

```python
x
```

```python
y
```

```python
lr=LinearRegression()
lr.fit(x,y)
```

```python
y_pred_lr=lr.predict(x)
y_pred_lr
```

```python
lr.score(x,y)
```

```python
#to perform the regularization techniques
lasso=Lasso(alpha=1000)
lasso.fit(x,y)
```

```python
y_pred_lasso=lasso.predict(x)
y_pred_lasso
```

```python
lasso.score(x,y)
```

```python
#l2 regularization
ridge=Ridge(alpha=1000)
ridge
```

```python
lr=lr.fit(x,y)
lr
```

```python
lr.score(x,y)
```

```python
#polynomial
from sklearn.preprocessing import PolynomialFeatures
poly=PolynomialFeatures(degree=2)
x_poly=poly.fit_transform(x)
x_poly
poily_model=LinearRegression()
poily_model.fit(x_poly,y)
y_pred_poly=poily_model.predict(x_poly)
y_pred_poly
poily_model.score(x_poly,y)
```
