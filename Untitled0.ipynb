```python
import pandas as pd
df=pd.read_csv('tested.csv')
df
```

```python
df.head()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt
sns.histplot(df['Age'],kde=True,bins=20)
plt.title('Age Distribution')
plt.show()
```

```python
#density plot
sns.kdeplot(data=df,x='Age',color='yellow',shade=True)
plt.title('Density plot of Age Distribution')
plt.show()
```

```python
#violinplot
sns.violinplot(x='Survived',y='Age',data=df,palette='coolwarm')
plt.title('Violin plot')
plt.show()
```

```python
#swarmplot
sns.swarmplot(x='Survived',y='Age',data=df,palette='coolwarm')
plt.title('Swarm plot')
plt.show()
```

```python
#heatmap
num_df=df.select_dtypes(include=["int64","float64"])
corr=num_df.corr()
plt.figure(figsize=(10,6))
sns.heatmap(corr,annot=True,cmap='coolwarm')
plt.title('Correlation Heatmap')
plt.show()
```

```python
num_df=df.select_dtypes(include=["int64","float64"])
corr=num_df.corr()
plt.figure(figsize=(10,6))
sns.heatmap(corr,annot=True,cmap='coolwarm')
plt.title('Heatmap')
plt.show()
```

```python
#Pair plot
cols=['Survived','Pclass','Age','Fare']
sns.pairplot(
    df[cols],
    hue='Survived',
    height=3,
    aspect=1.2

)
plt.show()
```
