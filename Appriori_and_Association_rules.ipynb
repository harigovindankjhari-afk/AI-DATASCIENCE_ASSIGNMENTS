```python
from mlxtend.frequent_patterns import apriori,association_rules
import warnings
warnings.filterwarnings('ignore')
```

```python
import pandas as pd
```

```python
df=pd.DataFrame([
    ['Milk','Bread','Butter'],
    ['Beer','Bread',None],
    ['Milk','Bread','Beer'],
    ['Bread','Butter',None],
    ['Milk','Butter',None]
    ])
df
```

```python
df=pd.get_dummies(df.stack()).groupby(level=0).sum()
```

```python
df
```

```python
frequent = apriori(df, min_support=0.1, use_colnames=True)
rules = association_rules(frequent, metric='confidence', min_threshold=0.22)
```
