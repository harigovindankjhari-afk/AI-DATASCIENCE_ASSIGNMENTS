```python
#Range,variance,sd
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

data = {
  'month' :['Jan','Feb','March','April','May','June','July','Aug','Sep','Oct','Nov','Dec'],
  'sales' :[200,220,210,240,260,270,265,280,300,320,350,500],
  'expense' :[120,140,130,150,160,170,168,175,180,190,200,260],
  'profit' :[80,80,80,90,100,100,97,105,120,120,150,240]
}
df=pd.DataFrame(data)
print("Dataset",df)
df
```

```python
#finding mean,range,variance,sd
print("MEAN-",df['profit'].mean())
print("RANGE-",df['profit'].max()-df['profit'].min())
print("VARIANCE-",df['profit'].var())
print("STANDARD DEVIATION-",df['profit'].std())
```

```python
#Quartile and TQR
Q1=df['profit'].quantile(0.25)
Q2=df['profit'].quantile(0.50)
Q3=df['profit'].quantile(0.75)

TQR = Q3-Q1
print('Q1-',Q1)
print('MEDIAN-',Q2)
print('Q3-',Q3)
print('TQR-',TQR)
```

```python
#outlier detection
lower=Q1-1.5*TQR
upper=Q1+1.5*TQR
print("LOWER-",lower)
print("UPPER-",upper)
outliers=df[(df['profit']<lower)|(df['profit']>upper)]
print("OUTLIERS-",outliers)
```

```python
#SKEWNESS AND KURTOSIS
print("SKEWNESS-",df['profit'].skew())
print("KURTOSIS-",df['profit'].kurt())
```
